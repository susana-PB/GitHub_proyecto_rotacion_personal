```python
# Ver cuántas variables categóricas hay
categoricas = df_original.select_dtypes(include=['object']).columns
categoricas

```

```python
# Análisis de cada variable categórica
for col in categoricas:
    print(f"🔹 {col}:")
    
    # Cuántos valores únicos tiene
    unicos = df_original[col].unique()
    print(f"   Valores posibles: {list(unicos)}")
    
    # Cuántos casos de cada valor
    conteo = df_original[col].value_counts()
    print(f"   Cantidad de cada valor:")
    
    for valor, cantidad in conteo.items():
        porcentaje = (cantidad / len(df_original)) * 100
        print(f"   • {valor}: {cantidad} personas ({porcentaje:.1f}%)")
    
    print("-" * 40)
```

```python
# Análisis Attrition 
print("\n ANÁLISIS DE ROTACIÓN (ATTRITION)")
print("-" * 40)

attrition_counts = df_original['Attrition'].value_counts()
print(f"Personas que NO se fueron: {attrition_counts.get('No', 0)}")
print(f"Personas que SÍ se fueron: {attrition_counts.get('Yes', 0)}")

tasa_rotacion = (attrition_counts.get('Yes', 0) / len(df_original)) * 100
print(f" Tasa de rotación: {tasa_rotacion:.1f}%")
print(f" Por cada 100 personas, {tasa_rotacion:.0f} se van")

# Ver cómo afecta OverTime a la rotación 
print("\n ¿AFECTAN LAS HORAS EXTRAS A LA ROTACIÓN?")
print("-" * 40)

# Crear una tabla simple
tabla_overtime = df_original.groupby('OverTime')['Attrition'].value_counts().unstack()
print(tabla_overtime)

# Calcular porcentajes
print("\nPorcentajes por grupo:")
for overtime in ['Yes', 'No']:
    total = tabla_overtime.loc[overtime].sum()
    se_fue = tabla_overtime.loc[overtime, 'Yes'] if 'Yes' in tabla_overtime.columns else 0
    porcentaje = (se_fue / total) * 100 if total > 0 else 0
    
    print(f"• Con horas extras ({overtime}): {porcentaje:.1f}% se van")

print("=" * 50)
print(" Análisis de variables categóricas completado")
```
