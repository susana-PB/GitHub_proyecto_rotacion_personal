### 1. Introducción y Contexto del Proyecto
Este notebook marca el inicio de mi Análisis Exploratorio de Datos (EDA) sobre la retención de las personas y la satisfacción laboral, utilizando un dataset público de IBM HR Analytics.
Mi objetivo es demostrar cómo el análisis de datos puede pasar de ser una tarea descriptiva a una herramienta estratégica para la toma de decisiones en Recursos Humanos, enfocándome específicamente en el bienestar y la permanencia del talento.

### 2. El Problema de Negocio (Business Case)
En el sector tecnológico, la rotación de personal (attrition) es costosa y disruptiva. No solo implica el gasto de reclutamiento y formación de nuevo personal, sino también la pérdida de conocimiento institucional y la posible desmotivación de los equipos existentes.
La pregunta central de negocio que busco responder es:
*¿Qué factores específicos, medibles a través de los datos de RRHH, influyen más en la decisión de una persona de permanecer o marcharse de la empresa? ¿Podemos predecir el riesgo de rotación basándome en su nivel de satisfacción y bienestar?*
Mi análisis pretende ofrecer insights accionables que permitan a quienes lideran RRHH diseñar intervenciones dirigidas a mejorar la experiencia de las personas y, por ende, la retención.

### 4. Formulación de Hipótesis de Trabajo

Nota Metodológica: Conceptos como "burnout" o "cultura organizacional" no están medidos directamente en el dataset.
Para evaluar estas hipótesis, crearé variables proxy combinando indicadores disponibles:
- Burnout risk = Bajo equilibrio trabajo-vida + baja involucración laboral
- Cultura positiva = Alta satisfacción con ambiente + relaciones laborales
- Alto engagement = Alta involucración + buen desempeño
- Oportunidades desarrollo = Formación frecuente + tiempo razonable en puesto

#### *Hipótesis Principal*
La satisfacción de las personas influye directamente en su bienestar profesional y en su decisión de permanecer en la empresa.  
**Premisa:** Un ambiente laboral positivo, medido a través de variables proxy que crearé, fortalece el sentido de pertenencia y reduce la rotación.

#### *Hipótesis Secundarias*
- **H1: Satisfacción y Bienestar:** Las personas más satisfechas con su trabajo experimentan menos desgaste profesional (burnout).
- **H2: Satisfacción y Pertenencia:** Las personas satisfeches tienen menor probabilidad de dejar la empresa y permanecen más tiempo en ella.
- **H3: Compromiso y Equilibrio:** Las personas comprometidas con su trabajo y con un buen equilibrio entre vida personal y laboral muestran mayor satisfacción y menos desgaste.
- **H4: Acciones de Bienestar y ambiente laboral:** Las iniciativas de la empresa relacionadas con bienestar, desarrollo y reconocimiento influyen positivamente en la satisfacción y retención del personal.
- **H5: Carga Laboral y Rotación:** Las personas que realizan horas extras frecuentes tienen mayor probabilidad de dejar la empresa.
